<h2>Travelling Salesperson Problem (Introduction)</h2>
In these notes, we will introduce the travelling salesperson problem (TSP). TSPs are a very important optimization problem over graphs. We will first define TSPs and motivate some of their applications. Next, we will prove that TSPs are NP-Complete. We will also prove that it is NP-Complete to approximate a general TSP to within a constant factor.

A TSP problem takes a <u>weighted and complete graph</u> that could be undirected or directed (as we will see below). The graph has  𝑛 vertices that we refer to by their indices  𝑉={1,…,𝑛}.
For every pair of vertices  𝑖,𝑗 where  𝑖≠𝑗, there must be an edge  (𝑖,𝑗) in the graph. In other words, we will work with <u>complete graphs</u> (i.e., every pair of vertices should be connected).
Each edge  (𝑖,𝑗) has an edge cost denoted by  𝑐𝑖𝑗.

In general, a TSP is defined over a directed graph. But many instances satisfy the symmetry property wherein the graph is undirected and  𝑐𝑖𝑗=𝑐𝑗𝑖.

Given a graph as described above, a <u> TSP problem seeks </u> a tour that starts at some arbitrary vertex, visits every other vertex of the graph exacty once and returns to the start point such that the total sum of all edge weights involved in the tour is as small as possible.

- It is important to emphasize that a valid TSP tour visits all vertices of the graph exactly once except for the starting point which may be regarded as being visited twice.
- The starting point of a TSP tour is arbitrary because it is a cycle.

<h2>Applications of TSP</h2>
There are many <u>types of TSP instances</u> and they have a wide variety of applications. For instance, the simplest application is in the area of <u>planning package deliveries</u>. A company such as Fedex has its vehicles visit hundreds or thousands of homes each day in a given area to deliver packages. It is important that the vehicles visit each of these homes exactly once and do so in a manner that minimizes the distance or time taken.
In fact, extensions of TSPs exist where there are more than one "salespersons" who visit the edges of the graph and the requirement is still that each node be visited exactly once by some salesperson. These are called <u>vehicle routing problems</u>.

TSPs are used in manufacturing to <u>schedule the movement of a drill bit used to manufacture printed circuit boards (PCB)</u>. Here, the PCB has many locations where one needs to have holes during the manufacturing process. Optimizing the movement of the arm to visit each of these sites can save manufacturing time and energy.

TSP-like problems have applications to <u>robotics</u> and other areas of engineering as well.

<h2>NP-Completeness of TSP</h2>

We already saw a proof that solving the (decision version) of a TSP is NP-complete in a previous course. Recall the decision version of a TSP:

<b>Input</b> A TSP instance: graph and edge weights along with a limit  𝐾.
<b>Output</b> Does there exist a TSP tour whose total cost is  ≤𝐾?

As we saw in a previous course, this can be shown NP-complete by first proving that it belongs to NP: a proposed solution can be checked easily by checking that it is indeed a tour and its total cost is <= K. To prove NP-completeness, we reduce from Hamiltonian Cycle problem.

The Hamiltonian Cycle problem is as follows: given an undirected graph, is there a cycle that visits all the nodes in the graph precisely once?

Note the relationship to a TSP. A TSP tour is in fact a Hamiltonian cycle. However, TSPs are solved over complete graphs where there is no doubt that there exists a Hamiltonian cycle. In fact, <u> there exists  𝑛! Hamiltonian cycles </u> in a complete graph of size  𝑛 . The key question in TSP is to find a tour with low total cost. In contrast, the graph for a Hamiltonian Cycle problem is not a complete graph. It is missing many edges and thus the question of whether or not the graph has a Hamiltonian cycle becomes key.










Richard Karp in his great 1972 paper proved that Hamiltonian cycle problem is NP-complete using a very clever reduction from 3-CNF-SAT:
Richard M. Karp (1972). Reducibility Among Combinatorial Problems. In R. E. Miller; J. W. Thatcher (eds.). Complexity of Computer Computations. New York: Plenum. pp. 85–103.

We can easily reduce Hamiltonian Cycle Problem (HCP) to TSP (i.e., change graph so that if TSP has solution for changed graph then HCP also has solution for the original graph - note - change should be polynomial time algorithm) as follows:

Take the given graph for Hamiltonian cycle and give all the edges present in the graph a weight of  1.
Add all the missing edges in the graph (remember TSPs are solved over complete graphs) but give them a weight of  2.
Now we have a graph whose edges have weight  1 if they were present in the original input graph or weight  2
otherwise. We now ask if there exists a TSP tour whose cost is  ≤𝑛 in this graph. There are two possibilities:

There is a TSP tour of size  ≤𝑛. However, a TSP tour has  𝑛 edges in it and thus such a tour must have edge cost  1 for all its edges. Hence, this tour indicates a Hamiltonian cycle in the original graph.
On the other hand, if there is no such tour, the least cost tour must have total cost  𝑛+1
  or more. This immediately indicates that there are no Hamiltonian cycles in the original graph.
Thus, from the answer to the TSP problem, we constructed answers the Hamiltonian cycle problem in a direct manner. As a result, TSP is NP-complete.

In fact, a <u>simple modification of the construction above allows us to conclude that TSPs in general cannot be approximated to within a constant factor</u>.


<h2>Hardness of Approximating TSPs</h2>
Are there efficient algorithms that can approximate a TSP instance to within a constant factor  𝛼>1 . In other words, we seek an algorithm that in polynomial time finds a tour of cost  𝑇  with the guarantee that
OPT ≤ 𝑇 ≤ 𝛼 OPT
where OPT denotes the cost of the optimal tour. We show that unless P=NP, such an algorithm cannot exist.

Let us suppose for contradiction, such an an efficient algorithm exists with some approximation factor  𝛼>1
 . We will prove that using this algorithm, we can solve the Hamiltonian cycle problem quite efficiently (or in technical terms, <u> we reduce from the problem of Hamiltonian cycle to that of approximating TSPs </u>).

To do so, we modify the reduction from Hamiltonian cycle that we just presented, as follows:

As before, if an edge exists in the original graph assign it weight  1, as previouly described.However, for edges not in the original graph, we assign them with weight  (𝛼−1)𝑛+2
  where  𝛼
  is the approximation ratio of the supposed constant-factor approximation algorithm. 
This is the main change from before. For instance, if  𝛼=2, we give each of these edges a cost of  𝑛+2.

Now we ask if the TSP has a tour of total cost  ≤𝑛. If such a tour exists, then we conclude that the original problem has a Hamiltonian cycle.

On the other hand, if the original problem does not have a Hamiltonian cycle, we conclude that the least cost Hamiltonian tour must have cost at least  (𝛼−1)𝑛+2+(𝑛−1) because in the best case we can take  𝑛−1
edges of cost  1 but we are forced to take at least one edge of cost  (𝛼−1)𝑛+2. Therefore, the cost of the tour must be at least  𝛼𝑛+1.



Therefore, even though the approximation algorithm does not compute an optimal tour, its approximation guarantees are sufficient for us to resolve whether or not the graph has a Hamiltonian cycle by simply checking if the approximation algorithm finds tour of cost ≤𝛼𝑛 or not.

This construction makes it clear that <u> general TSPs cannot even be approximated to within a constant factor</u>.
