```python
from pulp import * # get all the definitions from pulp library 

def solve_problem_1():
    prob = LpProblem('Question1', LpMaximize)
    # create decision variables
    dvars = [LpVariable(f'w_{i}', cat='Integer',lowBound=-15, upBound=15) for i in range(1,6)] # declare variables as binary
    # if we declared variables as binary they are automatically treated as either 0 or 1 in the optimization 
    # create objective
    prob += 2*dvars[0] - 3*dvars[1] + dvars[3]
    prob += dvars[0] - dvars[1] + dvars[2] <= 5
    prob += dvars[0] - dvars[1] + 4*dvars[2] <= 7
    prob += dvars[0] + 2*dvars[1] - dvars[2] + dvars[3] <= 14
    prob += dvars[2] - dvars[3] + dvars[4] <= 7
    status = prob.solve()
    assert status == constants.LpStatusOptimal
    print('Success: optimal answer found')
    print(dvars)
    solution_values = [(x, x.varValue) for x in dvars]
    opt_val_dict = {}
    for (name, val) in solution_values:
        opt_val_dict[f'{name}'] = val
        print(f'\t Value of # {name}: {val} ')
    print(f'opt_val_dict={opt_val_dict}')
    opt_val = 2*opt_val_dict['w_1'] - 3*opt_val_dict['w_2'] + opt_val_dict['w_3']
    print(f'optimum value={opt_val}')
solve_problem_1()
```

```python
from pulp import * # get all the definitions from pulp library 

def solve_problem_2_ILP():
    prob = LpProblem('Question2_ILP', LpMaximize)
    # create decision variables
    dvars = [LpVariable(f'w_{i}', cat='Integer',lowBound=-1, upBound=1) for i in range(1,4)] # declare variables as binary
    # if we declared variables as binary they are automatically treated as either 0 or 1 in the optimization 
    # create objective
    prob += 2*dvars[0] - 3*dvars[1] + dvars[2]
    prob += dvars[0] - dvars[1] >= 0.5
    prob += dvars[0] - dvars[1] <= 0.75
    prob += dvars[1] - dvars[2] <= 1.25
    prob += dvars[1] - dvars[2] >= 0.95
    status = prob.solve()
    if status == constants.LpStatusInfeasible:
        print('Problem is infeasible')
        return
    elif status == constants.LpStatusUnbounded:
        print('Problem is unbounded. Cannot proceed')
        return
    else:
        assert status == constants.LpStatusOptimal, 'Something went wrong while solving since status is either undefined or unsolved'
        
solve_problem_2_ILP()

```
