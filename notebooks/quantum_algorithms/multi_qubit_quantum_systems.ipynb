<h2>Entangled Qubits</h2>
Not all two qubit states can be written as a direct product of super positions: this fact has enormous implications. Consider the famous Bell state:



The possibility of entanglements in quantum mechanics was first proposed by Einstein, Podolsky and Rosen as a thought experiment in order to show that the postulates of quantum mechanics can yield physically counter-intuitive consequences. They imagine two particles emitted simultaneously in opposite directions by a single atom that stays stationary after emitting them. If we measure the position of one of them precisely, then we know the position of the other. Likewise, if we measure the velocity (or momentum) of the other particle, we know the momentum of the first one. By measuring position of one of the particles and simultaneously the momentum of the other, we can seemingly "cheat" the uncertainty principle that says that position and momentum of a particle cannot be measured with accuracy at the same time (actually the uncertainty principle says something much more far reaching). To get around this contradiction, it was suggested that the two particles are entangled. Measurement of position on one, makes the momentum of the other photon completely uncertain.

Likewise, the Bell state represents an entanglement of the two qubits.

```python
# modified from IBM Qiskit tutorial: https://github.com/Qiskit/qiskit
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer

qc1 = QuantumCircuit(2)# A quantum circuit with a four qubits. Qubits are indexed starting from 0
qc1.h(0) # apply hadamard on qubit # 0
qc1.cx(0, 1)# apply control-X with 0 as the control bit and 1 as the bit being operated on
qc1.cp(np.pi/2, 1, 0)
backend = Aer.get_backend('statevector_simulator')
new_circuit = transpile(qc1, backend)
job = backend.run(new_circuit)
result = job.result()
statevector = result.get_statevector()
print(f'({statevector[0]:.2f}) |00> + ({statevector[1]:.2f}) |01> + ({statevector[2]:.2f}) |10> + ({statevector[3]:.2f}) |11> ')
print('Note: python uses j = sqrt(-1)')
qc1.draw('mpl')
```

```python
# We can have gates controlled by multiple control qubits. The most famous one is called the Tofolli gate, which is a control not gate but with two control qubits. A Tofolli gate operating on bit  𝑏3
#  with control qubits  𝑏1,𝑏2  negates  𝑏3  if  𝑏1,𝑏2  are both  |1⟩. If either of  𝑏1,𝑏2 are zero, it leaves  𝑏3 unchanged.

# modified from IBM Qiskit tutorial: https://github.com/Qiskit/qiskit
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer

qc1 = QuantumCircuit(3)# A quantum circuit with a three qubits. Qubits are indexed starting from 0
qc1.h(0) # apply hadamard on qubit # 0
qc1.cx(0, 1)# apply control-X with 0 as the control bit and 1 as the bit being operated on
qc1.ccx(0, 1, 2)
backend = Aer.get_backend('statevector_simulator')
job = backend.run(transpile(qc1, backend))
result = job.result()
statevector = result.get_statevector()
def three_qubit_state_vector_to_str(statevector):
    pure_states = [f'|{b1}{b2}{b3}>' for b1 in [0,1] for b2 in [0,1] for b3 in [0,1]]
    state_vectors = [f'{statevector[i]:.2f}' for i in range(8)]
    result_string = '+'.join([f'{s1}{s2}' for (s1, s2) in zip( state_vectors, pure_states)])
    return result_string
print(three_qubit_state_vector_to_str(statevector))
print('Note: python uses j = sqrt(-1)')
qc1.draw('mpl')
```

```python
# modified from IBM Qiskit tutorial: https://github.com/Qiskit/qiskit
import numpy as np
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer

#1. Create a circuit with 3 qubits
qc1 = QuantumCircuit(3)# A quantum circuit with a three qubits. Qubits are indexed starting from 0
#2. Apply a Hadamard to the first twwo qubits to create an uniform superposition 
qc1.h(0) # apply hadamard on qubit # 0
qc1.h(1) # apply a hadamard on qubit # 1

# apply not gates to the two inputs 
qc1.x(0)
qc1.x(1)
# apply a ccx gate and with bits 0, 1 with result bit 2
qc1.ccx(0, 1, 2)
# negate back the inputs to restore them
qc1.x(0)
qc1.x(1)
# negate the result
qc1.x(2)
backend = Aer.get_backend('statevector_simulator')
job = backend.run(transpile(qc1, backend))
result = job.result()
statevector = result.get_statevector()
def three_qubit_state_vector_to_str(statevector):
    pure_states = [f'|{b3}{b2}{b1}>' for b1 in [0,1] for b2 in [0,1] for b3 in [0,1]]
    state_vectors = [f'{statevector[i]:.2f}' for i in range(8)]
    result_string = ' + '.join([f'{s1}{s2}' for (s1, s2) in zip( state_vectors, pure_states)])
    return result_string
print(three_qubit_state_vector_to_str(statevector))
print('Note: python uses j = sqrt(-1)')
qc1.draw('mpl')

```

Question: Can we implement a quantum XOR gate that maps  |𝑣1𝑣2⟩ → |𝑣1(𝑣1⊕𝑣2)⟩?

As it turns out, yes. It is a gate that you have seen before: the controlled-X gate. Also, XOR operator does not need a result qubit since it is unitary.

Consider a classical circuit made up of  𝑘>0 logic gates AND, OR and NOT that implements a Boolean function  𝑓(𝑏1,…,𝑏𝑚)
  for inputs  𝑏1,…,𝑏𝑚. Any Boolean function can be realized in this way.



We can build an equivalent quantum circuit with input qubits  𝑏1,…,𝑏𝑚, a result qubit  𝑟 and ancillary qubits  𝑞1,…,𝑞𝑘
  (as many as the number of logic gates) that implements the operation

  

  

```python
import numpy as np
from qiskit import QuantumCircuit, QuantumRegister

def apply_qor2(q_circuit, bit1_idx, negate_bit1, bit2_idx, negate_bit2, ancilla_idx):
    if not negate_bit1:
        q_circuit.x(bit1_idx)
    if not negate_bit2:
        q_circuit.x(bit2_idx)
    q_circuit.ccx(bit1_idx, bit2_idx, ancilla_idx)
    if not negate_bit1:
        q_circuit.x(bit1_idx)
    if not negate_bit2:
        q_circuit.x(bit2_idx)
    q_circuit.x(ancilla_idx)
    
def apply_qand2(q_circuit, bit1_idx, bit2_idx, ancilla_idx ):
    q_circuit.ccx(bit1_idx, bit2_idx, ancilla_idx)

    

#1. Create a circuit with 3 input qubits + three ancillary qubits + one result qubit
inputs = QuantumRegister(3,'b') # input register = 3 qubits
ancilla = QuantumRegister(3, 'q') # ancillary qubits = 3 qubits
res = QuantumRegister(1, 'r') # result = 1 qubit

qc = QuantumCircuit(inputs, ancilla, res)# A quantum circuit with a sevem qubits. Qubits are indexed starting from 0

apply_qor2(qc, inputs[0], False, inputs[1], True, ancilla[0]) # apply b1 \/ ! b2 with result qubit index 3
apply_qor2(qc, inputs[0], False, inputs[1], False, ancilla[1]) # b1 \/ b2 with result quibit 4
apply_qor2(qc, ancilla[1], False, inputs[2], True, ancilla[2]) # previous result or !b3
apply_qand2(qc, ancilla[0], ancilla[2], res[0])
qc.draw('mpl', style="iqp")

```

<h2>Quantum Parallelism</h2>



Example
Quantum parallelism is achieved by evaluating the circuit above using the uniform superposition for the input qubit, which is in turn obtained by using the Hadamard gate on them. Suppose we measure the result qubit, then we either obtain a  |0⟩ for the result in which case, the super position collapses to all possible values of  |𝑏1𝑏2𝑏3⟩ such that  𝑓(𝑏1,𝑏2,𝑏3)=0. Similar process happens if the result is  |1⟩.



```python
import numpy as np
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit.visualization import plot_histogram, plot_state_city
from qiskit_aer import Aer

## Copy and paste the circuit from previous example
def apply_qor2(q_circuit, bit1_idx, negate_bit1, bit2_idx, negate_bit2, ancilla_idx):
    if not negate_bit1:
        q_circuit.x(bit1_idx)
    if not negate_bit2:
        q_circuit.x(bit2_idx)
    q_circuit.ccx(bit1_idx, bit2_idx, ancilla_idx)
    if not negate_bit1:
        q_circuit.x(bit1_idx)
    if not negate_bit2:
        q_circuit.x(bit2_idx)
    q_circuit.x(ancilla_idx)
    
def apply_qand2(q_circuit, bit1_idx, bit2_idx, ancilla_idx ):
    q_circuit.ccx(bit1_idx, bit2_idx, ancilla_idx)


def create_uniform(q_circuit, bits):
    for i in bits:
        q_circuit.h(i) # apply Hadamard gate

#1. Create a circuit with 3 input qubits + three ancillary qubits + one result qubit
inputs = QuantumRegister(3,'b') # input register = 3 qubits
ancilla = QuantumRegister(3, 'q') # ancillary qubits = 3 qubits
res = QuantumRegister(1, 'r') # result = 1 qubit
res_measured = ClassicalRegister(1, 'z') # hold the measurement output
qc = QuantumCircuit(inputs, ancilla, res, res_measured)# A quantum circuit with a sevem qubits. Qubits are indexed starting from 0
create_uniform(qc, [inputs[0], inputs[1], inputs[2]]) # apply hadamard gate
apply_qor2(qc, inputs[0], False, inputs[1], True, ancilla[0]) # apply b1 \/ ! b2 with result qubit index 3
apply_qor2(qc, inputs[0], False, inputs[1], False, ancilla[1]) # b1 \/ b2 with result quibit 4
apply_qor2(qc, ancilla[1], False, inputs[2], True, ancilla[2]) # previous result or !b3
apply_qand2(qc, ancilla[0], ancilla[2], res[0])
qc.measure(res, res_measured)

qc.draw('mpl', style="iqp")
```

```python
# Transpile for simulator
simulator = Aer.get_backend('aer_simulator')
circ = transpile(qc, simulator)

# Run and get counts
result = simulator.run(qc).result()
counts = result.get_counts(circ)
plot_histogram(counts, title='result counts (1000 simulations)')
```
