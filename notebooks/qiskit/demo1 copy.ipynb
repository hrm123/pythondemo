```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
%matplotlib inline

num_qubits = 4
num_classic_registers = 0 # store the result of measurement
qc = QuantumCircuit(num_qubits)

# apply Hadamard gate on qubit whose initial state is |0>
qc.h(0)
#statevector = |0000> + |0001>
qc.cx(0,1)
#statevector = |0000> + |0011>
qc.cx(0,2)
#statevector = |0000> + |0111>
qc.cx(0,3)
#statevector = |0000> + |1111>
qc.save_statevector()
qc.measure_all()
qc.draw('mpl', style="iqp")

# to run this quantum program on quantum computer we will need to go to research lab. Hence IBM gives quantum computer in cloud. 
simulator_aer = AerSimulator()
qc_aer = transpile(qc, backend=simulator_aer)
result = simulator_aer.run(qc_aer).result()
state_1 = result.get_statevector()
state_1.draw('latex') # you can see output statevector by changing qc.save_statevector() to diffirent lines

```
