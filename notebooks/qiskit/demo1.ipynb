```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector

num_qubits = 2
num_classic_registers = 1 # store the result of measurement
qc = QuantumCircuit(num_qubits, num_classic_registers)

# apply Hadamard gate on qubit whose initial state is |0>
qc.h(1)
qc.measure(1,0) #qubit specifier, classicbit specifier
qc.h(0).c_if(0,1) # apply hadamard on qubit0 only if classicalbit 0 equals clasical 1
qc.save_statevector() # will show as dotted line in network diagram. The statevector object will be added to qc
qc.draw('mpl', style="iqp")
# to run this quantum program on quantum computer we will need to go to research lab. Hence IBM gives quantum computer in cloud. 
simulator_aer = AerSimulator()
qc_aer = transpile(qc, backend=simulator_aer)
result = simulator_aer.run(qc_aer).result()
state_1 = result.get_statevector()
state_1.draw('latex') # you will see state_1 would be 50% of time =  |1> tensor_product (1/sqrt(2)) ( |0> +|1>) =  (1/sqrt(2)) ( |10> +|11>) & 50% of time you will get |00>

```
