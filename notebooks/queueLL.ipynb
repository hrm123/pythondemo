```python
import collections
import queue as q
from multiprocessing import Queue

class QueueUsingDeque:
    def __init__(self):
        self.TwoEndedQueue =  collections.deque([])
    def enqueue(self,value): # append to end of the queue
        self.TwoEndedQueue.append(value)
    def dequeue(self): # remove from start of the queue
        return self.TwoEndedQueue.popleft()
    

class QueueUsingQueue:
    def __init__(self):
        self.Q =  q.Queue(maxsize=100)
    def enqueue(self,value): # append to end of the queue
        self.Q.put(value)
    def dequeue(self): # remove from start of the queue
        return self.Q.get()
    

customQueue = Queue(maxsize=100)
customQueue.put(1)
customQueue.get(1)

```

```python
queue = QueueUsingDeque()
queue.enqueue(1)
queue.enqueue(2)
queue.enqueue(3)
assert queue.dequeue()==1, "test1 failed"
queue.enqueue(4)
assert queue.dequeue()==2, "test2 failed"
print("all tests passed")

q = QueueUsingQueue()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
assert q.dequeue()==1, "test1 failed"
q.enqueue(4)
assert q.dequeue()==2, "test2 failed"
print("all tests passed")
```
